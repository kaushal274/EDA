```python
def storage_features(value):

    hdd = 0
    ssd = 0
    flash = 0
    hybrid = 0

    pattern = r'(\d+(?:\.\d+)?)\s*(TB|GB)\s*(SSD|HDD|Flash Storage|Hybrid)'

    matches = re.findall(pattern, value, flags=re.IGNORECASE)

    for size, unit, storage_type in matches:

        size = float(size)

        if unit.upper() == 'TB':
            size = size * 1024

        if storage_type.upper() == 'HDD':
            hdd += size

        elif storage_type.upper() == 'SSD':
            ssd += size

        elif storage_type.lower() == 'flash storage':
            flash += size

        elif storage_type.lower() == 'hybrid':
            hybrid += size

    return pd.Series([hdd, ssd, flash, hybrid])
```

```python
import re
```

```python
df[['HDD', 'SSD', 'Flash_Storage', 'Hybrid']] = df['Memory'].apply(storage_features)
```
